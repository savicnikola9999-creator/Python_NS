4. What is the most optimal skill to learn for Data Analysts? *(High Demand AND High Paying)*




```python
# Importing Libraries
import ast
import pandas as pd
import seaborn as sns
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
#dataset = load_dataset('lukebarousse/data_jobs')
#df = dataset['train'].to_pandas()


df = pd.read_csv(r"C:\Users\Nikola S\Desktop\Luke B\job_postings_flat.csv")

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

# Roles to explore

```python
df_plot = df['job_title_short'].value_counts().to_frame()

sns.set_theme(style='ticks')
sns.barplot(data=df_plot, x='count', y='job_title_short', hue='count', palette='dark:b_r', legend=False)
sns.despine()
plt.title('Number of Jobs per Job Title')
plt.xlabel('Number of Jobs')
plt.ylabel('')
plt.show()
```

# Roles to explore in Serbia

```python
df_plot = df[df["job_country"] == "Serbia"]
df_plot = df_plot['job_title_short'].value_counts().to_frame()

sns.set_theme(style='ticks')
sns.barplot(data=df_plot, x='count', y='job_title_short', hue='count', palette='dark:b_r', legend=False)
sns.despine()
plt.title('Number of Jobs per Job Title in Serbia')
plt.xlabel('Number of Jobs')
plt.ylabel('')
plt.show()
```

# Countries to explore

```python
df_plot = df['job_country'].value_counts().to_frame().head(15)

sns.set_theme(style='ticks')
sns.barplot(data=df_plot, x='count', y='job_country', hue='count', palette='dark:b_r', legend=False)
sns.despine()
plt.title('Number of Jobs per Country for Top 15')
plt.xlabel('Number of Jobs')
plt.ylabel('')
plt.show()
```

# Companies to explore

```python
df_plot = df["company_name"].value_counts().to_frame().sort_values(by="count", ascending=False)[3:].head(15)

sns.set_theme(style="ticks")
sns.barplot(data=df_plot, x="count", y="company_name", hue="count", legend=False)
sns.despine()
plt.ylabel("")
plt.xlabel("Count of jobs postnig")
plt.title("Jobs posting by companise")

plt.show()
```
